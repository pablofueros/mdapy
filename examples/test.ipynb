```python
import numpy as np
import numpy.typing as npt
```

```python
from mdapy.mda_dev import DataColumn

pts = DataColumn([1, 2, 3, 4, 5, 6, 7, 8])
theta = DataColumn([180, 90, 45, 30, 0, -30, -45, -90], directional=True)

columns = [pts, theta]
n_clusters = 3
seed_index = 0
```

```python
# Save the indices of the directional columns
dir_indices = np.where([col.directional for col in columns])[0]

dir_indices
```

```python
# Initialize the data matrix and the subset indices
matrix = np.array([col.np_array for col in columns]).T
subset_indices = np.zeros(n_clusters, dtype=np.int32)
next_index = seed_index

matrix
```

```python
# Initialize the distance matrix for caching all pairwise distances
dist_matrix = np.zeros((matrix.shape[0], matrix.shape[0]))

dist_matrix
```

```python
def distance(
    point1: npt.NDArray,
    point2: npt.NDArray,
    dir_indices: npt.NDArray,
    std_list: list[float],
):  # TODO: Return type !!!
    vector = np.zeros_like(point1)
    for ind, (p1_ind, p2_ind) in enumerate(zip(point1, point2)):
        if dir_indices is not None and ind in dir_indices:
            dist = ((p1_ind - p2_ind) + 180) % 360 - 180
        else:
            dist = p1_ind - p2_ind
        vector[ind] = dist / (np.sqrt(2) * std_list[ind])
    return np.linalg.norm(vector)
```
